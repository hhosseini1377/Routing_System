```python
import pickle
import re

import numpy as np

# CJK Unified Ideographs + Extension A (typical Chinese/Japanese/Kanji in prompts)
_CJK_RE = re.compile(r"[\u3400-\u4dbf\u4e00-\u9fff]")


def _has_cjk(s: str) -> bool:
    return bool(_CJK_RE.search(s))


with open('datasets/routerbench_raw.pkl', 'rb') as f:
    data = pickle.load(f)

# print(np.mean(data['meta/llama-2-70b-chat']))
print(data.keys())
print(len(data))
# Remove entries whose key is a string containing CJK (Chinese, etc.)
data = {
    k: v
    for k, v in data.items()
    if not (isinstance(k, str) and _has_cjk(k))
}
print(data['performance'][0])
# print(len(data))
for key in data.keys():
    if 'model_response' not in data[key] and 'total_cost' not in data[key]:
        try:
            print(f'{key}: {np.mean(data[key])}')
        except:
            pass




```
